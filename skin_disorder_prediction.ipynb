```python
# ============================================================
# 📚 IMPORT LIBRARIES
# ============================================================

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import RandomForestClassifier

from sklearn.svm import SVC

from xgboost import XGBClassifier

from catboost import CatBoostClassifier

print("✅ Libraries imported successfully!")
```

#CELL 4 — DATASET UPLOAD

```python
# ============================================================
# 📤 UPLOAD DATASET
# ============================================================

print("="*70)
print("📤 Upload Dermatology CSV Dataset")
print("="*70)

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("\n✅ Dataset loaded successfully!")

print("\n📊 Dataset Shape:", df.shape)
```

#CELL 5 — DATASET PREVIEW

```python
# ============================================================
# 📊 DATASET PREVIEW
# ============================================================

print("="*70)
print("🩺 DERMATOLOGY DATASET PREVIEW")
print("="*70)

display(df.head())

print("\n📋 Column Names:\n")

for col in df.columns:

    print(col)
```

#CELL 6 — BASIC DATA ANALYSIS

```python
# ============================================================
# 📊 BASIC DATA ANALYSIS
# ============================================================

print("="*70)
print("📊 DATASET INFORMATION")
print("="*70)

print("\n🔹 Shape of Dataset:")
print(df.shape)

print("\n🔹 Missing Values:")
print(df.isnull().sum())

print("\n🔹 Data Types:")
print(df.dtypes)

print("\n🔹 Statistical Summary:")
display(df.describe())
```

#CELL 7 — TARGET CLASS ANALYSIS

```python
# ============================================================
# 🎯 TARGET CLASS DISTRIBUTION
# ============================================================

target_col = df.columns[-1]

print("🎯 Target Column:", target_col)

plt.figure(figsize=(8,5))

sns.countplot(
    x=df[target_col],
    palette='viridis'
)

plt.title(
    '🩺 Skin Disorder Class Distribution',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Disease Class')

plt.ylabel('Count')

plt.show()

print("\n📊 Class Counts:\n")

print(df[target_col].value_counts())
```

#CELL 8 — MISSING VALUE HANDLING

```python
# ============================================================
# 🧹 HANDLE MISSING VALUES CORRECTLY
# ============================================================

print("="*70)
print("🧹 HANDLING MISSING VALUES")
print("="*70)

# ------------------------------------------------------------
# REPLACE '?' WITH NaN
# ------------------------------------------------------------

df.replace('?', np.nan, inplace=True)

# ------------------------------------------------------------
# CONVERT ALL COLUMNS TO NUMERIC WHERE POSSIBLE
# ------------------------------------------------------------

for col in df.columns:

    df[col] = pd.to_numeric(
        df[col],
        errors='ignore'
    )

# ------------------------------------------------------------
# CHECK MISSING VALUES
# ------------------------------------------------------------

print("\n🔍 Missing Values Before Cleaning:\n")

print(df.isnull().sum())

# ------------------------------------------------------------
# FILL NaN VALUES
# ------------------------------------------------------------

for col in df.columns:

    if df[col].dtype != 'object':

        df[col].fillna(
            df[col].median(),
            inplace=True
        )

# ------------------------------------------------------------
# FINAL CHECK
# ------------------------------------------------------------

print("\n✅ Missing values handled successfully!")

print("\n🔍 Missing Values After Cleaning:\n")

print(df.isnull().sum())
```

#CELL 9 — CORRELATION HEATMAP

```python
# ============================================================
# 🔥 CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(16,12))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    cmap='coolwarm'
)

plt.title(
    '🔥 Feature Correlation Heatmap',
    fontsize=16,
    fontweight='bold'
)

plt.show()
```

#CELL 10 — FEATURE & TARGET SPLIT

```python
# ============================================================
# 🎯 FEATURE & TARGET SPLIT
# ============================================================

target_col = df.columns[-1]

X = df.drop(target_col, axis=1)

y = df[target_col]

print("✅ Features Shape:", X.shape)

print("✅ Target Shape:", y.shape)
```

#CELL 11 — LABEL ENCODING

```python
# ============================================================
# 🔠 LABEL ENCODING
# ============================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("✅ Target labels encoded successfully!")

print("\n📌 Disease Classes:\n")

for i, cls in enumerate(label_encoder.classes_):

    print(f"{i} → {cls}")
```

#CELL 12 — TRAIN TEST SPLIT

```python
# ============================================================
# ✂️ TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y_encoded,

    test_size=0.2,

    random_state=42,

    stratify=y_encoded
)

print("✅ Training Shape:", X_train.shape)

print("✅ Testing Shape :", X_test.shape)
```

#CELL 13 — FEATURE SCALING

```python
# ============================================================
# ⚙️ FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("✅ Feature scaling completed!")
```

#CELL 14 — TRAIN LOGISTIC REGRESSION

```python
# ============================================================
# 🤖 LOGISTIC REGRESSION
# ============================================================

lr_model = LogisticRegression(max_iter=5000)

lr_model.fit(X_train_scaled, y_train)

lr_preds = lr_model.predict(X_test_scaled)

lr_acc = accuracy_score(y_test, lr_preds)

print("="*70)

print("✅ Logistic Regression Accuracy:", round(lr_acc*100,2), "%")
```

#CELL 15 — TRAIN RANDOM FOREST

```python
# ============================================================
# 🌲 RANDOM FOREST
# ============================================================

rf_model = RandomForestClassifier(

    n_estimators=200,

    random_state=42
)

rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_test)

rf_acc = accuracy_score(y_test, rf_preds)

print("="*70)

print("✅ Random Forest Accuracy:", round(rf_acc*100,2), "%")
```

#CELL 16 — TRAIN SVM

```python
# ============================================================
# ⚡ SUPPORT VECTOR MACHINE
# ============================================================

svm_model = SVC(

    probability=True,

    kernel='rbf'
)

svm_model.fit(X_train_scaled, y_train)

svm_preds = svm_model.predict(X_test_scaled)

svm_acc = accuracy_score(y_test, svm_preds)

print("="*70)

print("✅ SVM Accuracy:", round(svm_acc*100,2), "%")
```

#CELL 17 — TRAIN XGBOOST

```python
# ============================================================
# 🚀 XGBOOST CLASSIFIER
# ============================================================

xgb_model = XGBClassifier(

    objective='multi:softmax',

    num_class=len(np.unique(y_encoded)),

    eval_metric='mlogloss',

    random_state=42
)

xgb_model.fit(X_train, y_train)

xgb_preds = xgb_model.predict(X_test)

xgb_acc = accuracy_score(y_test, xgb_preds)

print("="*70)

print("✅ XGBoost Accuracy:", round(xgb_acc*100,2), "%")
```

#CELL 18 — TRAIN CATBOOST

```python
# ============================================================
# 🐱 CATBOOST CLASSIFIER
# ============================================================

cat_model = CatBoostClassifier(

    verbose=0,

    random_state=42
)

cat_model.fit(X_train, y_train)

cat_preds = cat_model.predict(X_test)

cat_preds = cat_preds.astype(int)

cat_acc = accuracy_score(y_test, cat_preds)

print("="*70)

print("✅ CatBoost Accuracy:", round(cat_acc*100,2), "%")
```

#CELL 19 — MODEL COMPARISON

```python
# ============================================================
# 📊 MODEL COMPARISON
# ============================================================

results = pd.DataFrame({

    'Model': [

        'Logistic Regression',

        'Random Forest',

        'SVM',

        'XGBoost',

        'CatBoost'
    ],

    'Accuracy': [

        lr_acc,

        rf_acc,

        svm_acc,

        xgb_acc,

        cat_acc
    ]
})

results = results.sort_values(

    by='Accuracy',

    ascending=False
)

display(results)

# ============================================================
# VISUALIZATION
# ============================================================

plt.figure(figsize=(10,5))

sns.barplot(

    data=results,

    x='Model',

    y='Accuracy',

    palette='viridis'
)

plt.title(

    '🏆 Model Accuracy Comparison',

    fontsize=16,

    fontweight='bold'
)

plt.ylim(0,1)

plt.show()

best_model_name = results.iloc[0]['Model']

print("\n🏆 BEST MODEL:", best_model_name)
```

#CELL 20 — CLASSIFICATION REPORT

```python
# ============================================================
# 📋 CLASSIFICATION REPORT
# ============================================================

print("="*70)

print("📋 CLASSIFICATION REPORT — BEST MODEL")

print("="*70)

if best_model_name == 'XGBoost':

    best_preds = xgb_preds

elif best_model_name == 'CatBoost':

    best_preds = cat_preds

elif best_model_name == 'Random Forest':

    best_preds = rf_preds

elif best_model_name == 'SVM':

    best_preds = svm_preds

else:

    best_preds = lr_preds

print(

    classification_report(

        y_test,

        best_preds
    )
)
```

#CELL 21 — CONFUSION MATRIX

```python
# ============================================================
# 🔥 CONFUSION MATRIX
# ============================================================

plt.figure(figsize=(10,7))

sns.heatmap(

    confusion_matrix(y_test, best_preds),

    annot=True,

    fmt='d',

    cmap='Purples'
)

plt.title(

    '🔥 Confusion Matrix',

    fontsize=16,

    fontweight='bold'
)

plt.xlabel('Predicted')

plt.ylabel('Actual')

plt.show()
```

#CELL 22 — FEATURE IMPORTANCE ANALYSIS

```python
# ============================================================
# 📊 FEATURE IMPORTANCE
# ============================================================

if best_model_name == 'Random Forest':

    importances = rf_model.feature_importances_

elif best_model_name == 'XGBoost':

    importances = xgb_model.feature_importances_

else:

    importances = cat_model.feature_importances_

feature_importance_df = pd.DataFrame({

    'Feature': X.columns,

    'Importance': importances
})

feature_importance_df = feature_importance_df.sort_values(

    by='Importance',

    ascending=False
)

display(feature_importance_df.head(15))

# ============================================================
# VISUALIZATION
# ============================================================

plt.figure(figsize=(10,8))

sns.barplot(

    data=feature_importance_df.head(15),

    x='Importance',

    y='Feature',

    palette='magma'
)

plt.title(

    '🧠 Top 15 Important Medical Features',

    fontsize=16,

    fontweight='bold'
)

plt.show()
```

#CELL 23 — REAL-TIME SKIN DISORDER PREDICTION

```python
# ============================================================
# 🎯 ADVANCED AI SKIN DISORDER PREDICTION SYSTEM
# ============================================================

print("="*85)
print("🩺 ADVANCED AI SKIN DISORDER PREDICTION SYSTEM")
print("="*85)

print("""
📌 Severity Scale

0 → Absent
1 → Mild
2 → Moderate
3 → Severe
""")

# ============================================================
# DISEASE NAME MAPPING
# ============================================================

disease_mapping = {

    0: 'Psoriasis',

    1: 'Seborrheic Dermatitis',

    2: 'Lichen Planus',

    3: 'Pityriasis Rosea',

    4: 'Chronic Dermatitis',

    5: 'Pityriasis Rubra Pilaris'
}

# ============================================================
# SYMPTOM LIST
# ============================================================

feature_labels = {

    'erythema': 'Skin Redness',

    'scaling': 'Skin Scaling',

    'itching': 'Itching',

    'koebner_phenomenon': 'Koebner Phenomenon',

    'polygonal_papules': 'Polygonal Papules'
}

# ============================================================
# TAKE BASE SAMPLE
# ============================================================

sample_input = X.sample(1).copy()

# ============================================================
# USER INPUTS
# ============================================================

print("\n🩺 Enter Patient Symptom Severity:\n")

for idx, (feature, label) in enumerate(feature_labels.items(), start=1):

    if feature in sample_input.columns:

        print(f"{idx}. {label}")

        val = int(input("Enter Severity (0-3): "))

        sample_input[feature] = val

        print()

# ============================================================
# MODEL PREDICTIONS
# ============================================================

results = []

# ------------------------------------------------------------
# LOGISTIC REGRESSION
# ------------------------------------------------------------

lr_probs = lr_model.predict_proba(

    scaler.transform(sample_input)

)[0]

lr_pred = np.argmax(lr_probs)

lr_conf = np.max(lr_probs) * 100

results.append([
    'Logistic Regression',
    lr_pred,
    lr_conf
])

# ------------------------------------------------------------
# RANDOM FOREST
# ------------------------------------------------------------

rf_probs = rf_model.predict_proba(sample_input)[0]

rf_pred = np.argmax(rf_probs)

rf_conf = np.max(rf_probs) * 100

results.append([
    'Random Forest',
    rf_pred,
    rf_conf
])

# ------------------------------------------------------------
# SVM
# ------------------------------------------------------------

svm_probs = svm_model.predict_proba(

    scaler.transform(sample_input)

)[0]

svm_pred = np.argmax(svm_probs)

svm_conf = np.max(svm_probs) * 100

results.append([
    'SVM',
    svm_pred,
    svm_conf
])

# ------------------------------------------------------------
# XGBOOST
# ------------------------------------------------------------

xgb_probs = xgb_model.predict_proba(sample_input)[0]

xgb_pred = np.argmax(xgb_probs)

xgb_conf = np.max(xgb_probs) * 100

results.append([
    'XGBoost',
    xgb_pred,
    xgb_conf
])

# ------------------------------------------------------------
# CATBOOST
# ------------------------------------------------------------

cat_probs = cat_model.predict_proba(sample_input)[0]

cat_pred = np.argmax(cat_probs)

cat_conf = np.max(cat_probs) * 100

results.append([
    'CatBoost',
    cat_pred,
    cat_conf
])

# ============================================================
# RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(

    results,

    columns=[
        'Model',
        'Predicted_Class',
        'Confidence'
    ]
)

# ============================================================
# CONVERT CLASS TO DISEASE NAME
# ============================================================

results_df['Predicted_Disease'] = results_df[
    'Predicted_Class'
].map(disease_mapping)

# ============================================================
# BEST MODEL
# ============================================================

best_row = results_df.iloc[
    results_df['Confidence'].idxmax()
]

best_model = best_row['Model']

best_disease = best_row['Predicted_Disease']

best_conf = best_row['Confidence']

# ============================================================
# DISPLAY RESULTS
# ============================================================

print("\n" + "="*85)

print("🏆 FINAL DIAGNOSIS RESULT")

print("="*85)

print(f"🧠 Best Model            : {best_model}")

print(f"🩺 Predicted Disorder    : {best_disease}")

print(f"📊 Confidence Score      : {best_conf:.2f}%")

print("="*85)

print("\n📋 Medical Interpretation:")

print(
    f"The AI system predicts that the patient most likely "
    f"shows symptoms associated with: {best_disease}"
)

# ============================================================
# MODEL COMPARISON TABLE
# ============================================================

display(results_df)

# ============================================================
# VISUALIZE CONFIDENCE
# ============================================================

plt.figure(figsize=(10,5))

sns.barplot(

    data=results_df,

    x='Model',

    y='Confidence',

    palette='magma'
)

plt.title(

    '📊 Model Confidence Comparison',

    fontsize=16,

    fontweight='bold'
)

plt.ylim(0,100)

plt.ylabel('Confidence (%)')

plt.show()
```
